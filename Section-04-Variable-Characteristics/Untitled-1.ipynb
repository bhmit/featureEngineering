```python
# encode cabin 

dic_cabin_reduced = {k: i for i, k in enumerate(df['cabin_reduced'].unique(),0)}
dic_cabin_reduced
```

```python
df['cabin_mapped_reduced'] = df['cabin'].map(dic_cabin_reduced)
X_train['cabin_mapped_reduced'] = X_train['cabin_reduced'].map(dic_cabin_reduced)
X_test['cabin_mapped_reduced'] = X_test['cabin_reduced'].map(dic_cabin_reduced)
```

```python
dic_cabin_not_reduced = {
    k:i for i,k in enumerate(df['cabin'].unique(),0)
}


df['cabin_mapped_not_reduced'] = df['cabin'].map(dic_cabin_not_reduced)
X_train['cabin_mapped_not_reduced'] = X_train['cabin'].map(dic_cabin_not_reduced)
X_test['cabin_mapped_not_reduced']= X_test['cabin'].map(dic_cabin_not_reduced)
df.head()
```
